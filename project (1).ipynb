```python
data.columns = data.columns.str.strip()
data.columns = data.columns.str.replace(" ", "_")
data.columns = data.columns.str.lower()

data.columns
```

```python
data = data.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
data["payment_status"] = data["payment_status"].str.lower()
data["coupon_code"] = data["coupon_code"].str.upper()
print(data["payment_status"].head())
print(data["coupon_code"].head())
```

```python
data["delivery_days"] = pd.to_numeric(data["delivery_days"], errors="coerce")
data["customer_rating"] = pd.to_numeric(data["customer_rating"], errors="coerce")
print(data["delivery_days"].head())
print(data["customer_rating"].head())
print(data[["delivery_days", "customer_rating"]].dtypes)
```

```python
print("Duplicates:", data.duplicated().sum())

data = data.drop_duplicates()

print("After remove:", data.shape)
```

```python
# check before filling
if "customer_rating" in data.columns:
    data["customer_rating"] = data["customer_rating"].fillna(data["customer_rating"].mean())

if "payment_status" in data.columns:
    data["payment_status"] = data["payment_status"].fillna("unknown")

if "delivery_days" in data.columns:
    data["delivery_days"] = data["delivery_days"].fillna(data["delivery_days"].median())

if "coupon_code" in data.columns:
    data["coupon_code"] = data["coupon_code"].fillna("no_coupon")

data.isnull().sum()
```

```python
data = data.replace("unknown", np.nan)
data = data.replace("NONE", np.nan)
data = data.replace("none", np.nan)

print(data.isnull().sum())
```

```python
data["customer_city"] = data["customer_city"].str.title()
print(data["customer_city"].head())
```

```python
data["discount_percent"] = pd.to_numeric(
    data["discount_percent"].astype(str).str.replace("%", "", regex=False).str.strip(),
    errors="coerce"
)
print(data["discount_percent"].head())
print(data["discount_percent"].dtype)
```

```python
data["support_calls"] = pd.to_numeric(
    data["support_calls"].astype(str).str.replace("calls", "", regex=False).str.strip(),
    errors="coerce"
)
print(data["support_calls"].head())
print(data["support_calls"].dtype)
```

```python
data["discount_percent"] = data["discount_percent"].fillna(data["discount_percent"].median())
data["support_calls"] = data["support_calls"].fillna(data["support_calls"].median())
data["return_status"] = data["return_status"].fillna("Unknown")
data["customer_city"] = data["customer_city"].fillna("Unknown")
data["order_priority"] = data["order_priority"].fillna("Unknown")
data["membership"] = data["membership"].fillna("Unknown")

print(data.isnull().sum())
```
